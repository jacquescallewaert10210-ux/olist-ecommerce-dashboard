## Plotly Graph Objects


```python
import plotly.graph_objects as go
```

```python
#create the figure
fig=go.Figure()
fig.show()
```

```python
#create a graphical object from your data

trace=go.Scatter(x=tips_df.total_bill,
                 y=tips_df.tip)

trace
```

```python
#add it to the figure
fig.add_trace(trace)
fig.show()
```

```python
#we need to add some parameters to our graphical object, especially the type of graph (by default lines for a Scatter type object)

fig=go.Figure()
fig.add_trace(go.Scatter(x=tips_df.total_bill,
                         y=tips_df.tip,
                         mode="markers"))

fig.show()
```

```python
#we can add several many traces on our figure object, and do some formatting

fig=go.Figure()
fig.add_trace(go.Scatter(x=tips_df.total_bill,
                         y=tips_df.tip,
                         mode="markers",
                         name="Real observations"))

fig.add_trace(go.Scatter(x=[1,50],
                         y=[0.5,8],
                         mode="lines",
                         name="Linear Regression"))


fig.update_layout(
    title="Linear Regression Tip vs Total Bill",
    xaxis_title="Total Bill",
    yaxis_title='Tip',
    showlegend=True
)
```

```python
gapminder = pd.read_csv("https://raw.githubusercontent.com/plotly/datasets/master/gapminderDataFiveYear.csv")

gapminder.head()
```

### SunBurst

- <span style="color:red;"><code>path=["continent", "country"]</code></span> – sets the rings: inner ring = continent, outer ring = country (each country sits inside its continent).


- <span style="color:red;"><code>values="pop"</code></span> – size of each slice = that country's population (bigger population = bigger slice).

```python

gap2007 = gapminder[gapminder["year"] == 2007].copy()

fig = px.sunburst(
    gap2007,
    path=["continent", "country"],
    values="pop",
    color="lifeExp",
    color_continuous_scale="RdYlGn",
    title="World Population by Continent & Country (2007) — Colored by Life Expectancy",
)
fig.update_layout(height=650, margin=dict(t=60, l=0, r=0, b=0))
fig.show()
```

```python

fig = px.scatter(
    gapminder,
    x="gdpPercap", y="lifeExp",
    animation_frame="year", animation_group="country",
    size="pop", color="continent", hover_name="country",
    log_x=True, size_max=55,
    range_x=[100, 100000], range_y=[20, 90],
    labels={"gdpPercap": "GDP per Capita (log scale, USD)", "lifeExp": "Life Expectancy (years)"},
    title="Gapminder — GDP per Capita vs. Life Expectancy (1952-2007)",
)
fig.update_layout(height=550)
fig.show()
```
